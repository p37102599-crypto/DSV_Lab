```python
import pandas as pd

sales = pd.Series([120, 150, 180, 210])
print(sales)

marks = pd.Series([72, 81, 69, 90])

print(marks.values)
print(marks.index)
```

```python
temperatures  = pd.Series([28,31,30,29],index = ['Mon','Tue','Wed','Thu'])
print(temperatures)
```

```python
population  = pd.Series([35,42,31],index = ['Delhi','Mumbai','Jaipur'])
print(population['Jaipur'])
```

```python
score = pd.Series([35,68,44,79,52])
print(score[score>50])
```

```python
subjects =pd.Series([76,83,91],index = ['Physics','Chemistry','Maths'])
print(subjects['Chemistry'])
```

```python
s = pd.Series(['Yash','Rudra','Amit','Arush'])
print(s)
dict2 = s.to_dict()
print(dict2)
```

```python
prices = {'Pen': 10, 'Notebook': 55, 'Bag': 700}
price_series = pd.Series(prices)
price_series.name = 'price_series'
print(price_series)
```

```python
data = {'A': 10, 'B': 20, 'C': 30}
s = pd.Series(data, index=['A','B','C','D'])
print(s)
```

```python
print(s.isnull())
```

```python
print(s.notnull())
```

```python
sales.name = 'Monthly Sales'
sales.index.name = 'Month'
print(sales)
```

```python
s1 = pd.Series([18, 45, 27, 61], index=['P','Q','R','S'])
print(s1[s1 > 30])
```

```python
s = pd.Series([500,600,700],index=['East','West','North'])
print('West' in s)
print(500 in s)
```

```python
city_data = {'Delhi':80,'Pune':65,'Kochi':70}
s3 = pd.Series(city_data,index =['Delhi','Kochi','Pune','Chennai'])
print(s3)
```

```python
s4 = pd.Series({'W': 5, 'X': 10, 'Y': 15}, index=['W', 'X', 'Y', 'Z'])
print(pd.isnull(s4))
```

```python
print(pd.notnull(s4))
```

```python
s5 = pd.Series([10, 20, 30], index=['A', 'B', 'C'])
s6 = pd.Series([1, 2, 3], index=['C', 'A', 'B'])
result = s5+s6
print(result)
```

```python
stock =pd.Series([25,40,18],index = ['Item1','Item2','Item3'])
stock_dict = stock.to_dict()
print(stock_dict)
stock2 = pd.Series(stock_dict,index =['Item1','Item2','Item3'])
print(stock2)
```
