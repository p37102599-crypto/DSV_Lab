#Tuple

```python
# create  a tuple
student = (101,"Sanjiv","CSE",8.9)
print(student)
#tuple using index
print("Student information in first location: ",student[0])
print(student[1])
#negative index
print("Example for negative index: ",student[-2])

```

```python
subjects=("DSA" , "DSV","OS","PYTHON","ML","CN")
print(len(subjects))
```

```python
#slicing
subjects =("DSV","DSA","Statistics","SQL","ML",)
print(subjects[1:3])
print(subjects[:5])
print(subjects[4:])
```

```python
#tuple immutable
students_info = ("chirag",21,"CSE",9.7)
students_info[3]=9.1
print(students_info)
#it is not changeble
```

```python
#Nested tuple
student=(101,"Rahul",(89,76,66))
print(student)
print(student[2])
print(student[2][1])
```

```python
#count function
marks = (75,76,78,79,65,64,67,98)
print("Time of occurence",marks.count(75))
```

```python
#SET
##unordered collection of items
## Mutable
## allows the diffrent data types
## it doesn't support the idx
## it gives output with unique values(NO duplication Allowed in set)
```

```python
# symbol for set { }
# creating a set with similar data items
marks={76,44,56,79,79,98,98,87,99}
print(marks)
print(type(marks))
```

```python
# create a empty set
a={}
a =set()
# check its type
print(type(a))
```

```python
# set doesn't suport the idx
students={"Amit","Pragyan","Ashi"}
print(students[1])
```

```python
#add an item in the set
students={"Amit","Pragyan","Sneha"}
students.add("Priya")
print(students)
# to  add multiple items
students.update(["Ram","Vani","Devi"])
print(students)
```

```python
# remove an item from the set
students={"Amit","Riya","Priya"}
students.remove("Priya")
print(students)
```

```python
students= {"Amit","Pragyan","Priya"}
students.discard("Pragyan")
print(students)
```

```python
# create a set and clear list of item
students_info={"Amit","Pragyan","Ashi"}
print(students_info)
students_info.clear()
print(students_info)
```

```python
# Set Operations
python_students={"Priyeranjan","Sakshi","Shivam","Riya"}
java_students={"Karan","Vansh","Yuvraj","Piyush","Neha"}

#for union operation by using set
result=python_students.union(java_students)
print("Union Operation:",result)

#for intersection operation by using set
result_intersection=python_students.intersection(java_students)
print("Intersection Operation:",result_intersection)

# to find difference with two set
res_diff=python_students.difference(java_students)
print(res_diff)
```

```python
# Convert list to set
stud_id=[101,102,103,102,103,104,105]
unique_stud_id=set(stud_id)
print(unique_stud_id)
uniq_roll_no=list(unique_stud_id)
print(uniq_roll_no)
```
